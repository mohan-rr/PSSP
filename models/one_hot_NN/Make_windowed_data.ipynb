```python
import numpy as np
```

```python
np_data = np.load('Thesis_Data/np_data.npz')

window = 15 #31 residue stretch

np_seq    = np.zeros((4614740, 31), dtype='int8')
np_labels = np.zeros(4614740, dtype='int8')
# 4614740 x 31 x 22 tensor to hold

idx = 0
for chain in np_data.files:
  seq = np_data[chain][0]
  mask = np_data[chain][1]
  DSSP = np_data[chain][2]
  seqlen = len(np_data[chain][0])
  
  if seqlen > 30:
      for i in range (seqlen):
          if mask[i]:
              if i > 14: 
                  if i < seqlen - 16:
                      np_seq[idx,:] = seq[i-window:i+window+1] 
                      np_labels[idx] = DSSP[i]
                      idx += 1
                  else:
                      np_seq[idx,:] = np.append(seq[i-window:], np.full(i-seqlen+window+1, fill_value = 21, dtype='int8'))
                      np_labels[idx] = DSSP[i]
                      idx += 1
              else:
                  np_seq[idx, :] = np.append(np.full(15-i, fill_value= 21, dtype = 'int8'), seq[0:i+window+1])
                  np_labels[idx] = DSSP[i]
                  idx += 1
```

```python
np.savez('Thesis_Data/np_windows.npz', np_seq = np_seq, np_labels = np_labels)
```
