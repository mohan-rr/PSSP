```python
np_trainingdata = np.load('np_data.npz')

EM = np.ones((27,9261))
TM = np.ones((27,27))
Ends = np.ones(27)

for entry in np_trainingdata.files:
    seq = np_trainingdata[entry][0] 
    SS_seq = np_trainingdata[entry][2]
    seqlen = len(seq)
    
    #first
    SS_index = SS_seq[0]*3**2 + SS_seq[1]*3 + SS_seq[2]
    aa_index = seq[0]*21**2 + seq[1]*21 + seq[2]

    EM[SS_index,aa_index] += 1
    Ends[SS_index] += 1

    #loop
    for i in range(1, seqlen-2):
        last_index = SS_index
        aa_index = seq[i]*21**2 + seq[i+1]*21 + seq[i+2]
        SS_index = SS_seq[i]*3**2 + SS_seq[i+1]*3 + SS_seq[i+2]

        TM[last_index, SS_index] +=1
        EM[SS_index, aa_index] +=1

    #last
    Ends[SS_index] += 1

#turn into freq.
EM = EM/EM.sum(axis=1, keepdims=True)
TM = TM/TM.sum(axis=1, keepdims=True)
Ends = Ends/Ends.sum()
```

```python
np.savez('HMM_3o_params.npz', TM = TM, EM = EM, Ends = Ends)
```
